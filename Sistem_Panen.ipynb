```python
%%writefile penghitung_panen.py
def hitung_total_panen(daftar_hasil_panen):
    return sum(daftar_hasil_panen)

def hitung_diskon(total_harga, persen_diskon):
    nilai_diskon = total_harga * (persen_diskon / 100)
    harga_akhir = total_harga - nilai_diskon
    return harga_akhir

def main():
    hasil_panen = [120, 95, 140, 110, 105]
    total = hitung_total_panen(hasil_panen)
    print(f"Total hasil panen: {total} kg")

    harga_per_kg = 5000
    total_harga = total * harga_per_kg
    harga_setelah_diskon = hitung_diskon(total_harga, 10)
    print(f"Harga setelah diskon 10%: Rp{harga_setelah_diskon:.0f}")

if __name__ == "__main__":
    main()
```
