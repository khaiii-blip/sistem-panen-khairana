```python
from getpass import getpass
import subprocess
import base64

username = input("Username GitHub: ")
token = getpass("GitHub Token: ")

auth = base64.b64encode(
    f"{username}:{token}".encode()
).decode()

subprocess.run([
    "git",
    "-c",
    f"http.extraHeader=Authorization: Basic {auth}",
    "push",
    "-u",
    "origin",
    "fitur-diskon"
])
```
